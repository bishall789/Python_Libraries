```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'depart' : ['SE','CE','Civil','Electronic'],
    'stu' : [98,97,100,90]
}

df = pd.DataFrame(data)
plt.figure(figsize=(5,5))

plt.bar(df['depart'],df['stu'],color='green', label="Num")
plt.title('Students per Department')
plt.xlabel("department")
plt.ylabel("Number of Students")
plt.legend()

for i , val in enumerate(df['stu']):
    plt.text(df['depart'][i], val+2, str(val), ha='center')

plt.show()




```
