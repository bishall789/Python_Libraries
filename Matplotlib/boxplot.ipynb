BOX PLOT

```python
import matplotlib.pyplot as plt
```

```python
import numpy as np

salaries = [30000, 35000, 40000, 45000, 50000, 120000]

Q1 = np.percentile(salaries, 25)
Q3 = np.percentile(salaries, 75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = [x for x in salaries if x < lower_bound or x > upper_bound]
print("Outliers:", outliers)
```

```python

plt.boxplot(salaries)
plt.title("Salary Distribution with Outlier")
plt.ylabel("Salary ($)")
plt.show()
```

```python
import matplotlib.pyplot as plt

dept_A = [30000, 35000, 40000, 45000, 50000, 120000]
dept_B = [32000, 36000, 40000, 42000, 48000, 150000]

# Create a simple box plot
plt.boxplot([dept_A, dept_B], labels=['Dept A', 'Dept B'])

# Add titles and labels
plt.title("Salary Distribution by Department")
plt.ylabel("Salary ($)")

# Show plot
plt.show()
```
