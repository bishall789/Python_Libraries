```python
import matplotlib.pyplot as plt
```

HISTOGRAM

```python
age = [45, 50, 60, 70, 75, 80, 85, 90, 95]

plt.hist(age, bins=10, edgecolor='Black')
plt.title("Age distibution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

Multiple Histograms (Comparison)

```python
scores_A = [60, 65, 70, 75, 80]
scores_B = [55, 60, 68, 72, 78]

plt.hist(scores_A, bins=5, alpha=0.6, label="Class A")
plt.hist(scores_B, bins=5, alpha=0.6, label="Class B")

plt.legend()
plt.show()

# alpha -> transparency
```
