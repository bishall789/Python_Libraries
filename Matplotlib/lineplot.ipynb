```python
import pandas as pd
import matplotlib.pyplot as plt
```

LINE PLOT

```python

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x,y)
plt.show()


```

```python

x = [1, 2, 3, 4]
y = [10, 20, 25, 30]

plt.plot(x,y, color='red',linestyle='-', marker='o',linewidth=1, label='Sales')
plt.title("Montly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()


plt.show()
```

Multiple Line Plots

```python
sales_2024 = [10, 20, 25, 30]
sales_2025 = [15, 18, 28, 35]

plt.plot(x, sales_2024, label="2024")
plt.plot(x, sales_2025, label="2025")

plt.title("Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(False)


plt.show()
```

Saving figure

```python

sales_2024 = [10, 20, 25, 30]
sales_2025 = [15, 18, 28, 35]

plt.plot(x, sales_2024, label="2024")
plt.plot(x, sales_2025, label="2025")

plt.title("Sales Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(False)

# //Saving figure
plt.savefig("sales_trend.png")


plt.show()
```
